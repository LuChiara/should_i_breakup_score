# Should I breakup score

The Should-I-breakup score is a quantitative method to help you understand if you should leave your partner or not. This can be applied not only to your romantic partner, but also to friends, colleagues, and you name it!

$$ score = \frac{\sum (\text{Ptrait} \cdot W_p)}{\sum (\text{Ntrait} \cdot W_n)} \cdot \text{baseline}$$



$\sum$ = sum

$ Ptrait $= Positive trait, scoring from 1 (weak positive) to 5 (strong positive)

$Ntrait$ = Negative trait, scoring from 1 (weak negative) to 5 (intensely negative)

$W_p$ = Weight of the positive trait from 0 to 1

$N_n$ = Weight of the negative trait from 0 to 1.

$\sum W_p = 1$, $\sum W_n = 1$. Each total positive and total negative weight needs to add up to 1

$baseline$ = A % of how happy you are in the relationship from 0 to 1

## Let's get started with a Reusable Function

```python
def calculate_score(positive_traits, negative_traits, baseline):
    # numerator: sum of Ptrait * Wp
    numerator = sum(score * weight for (score, weight) in positive_traits.values())
    # denominator: sum of Ntrait * Wn
    denominator = sum(score * weight for (score, weight) in negative_traits.values())
    
    # Avoid division by zero
    if denominator == 0:
        return "Error: Denominator is zero. Cannot compute score."
    # Avoid sum of weights not equal to 1
    total_positive_weight = sum(weight for (_, weight) in positive_traits.values())
    total_negative_weight = sum(weight for (_, weight) in negative_traits.values())
    if not (abs(total_positive_weight - 1.0) < 1e-6 and abs(total_negative_weight - 1.0) < 1e-6):
        return "Error: Weights must sum to 1 for both positive and negative traits."

    score = (numerator / (1 + denominator)) * baseline
    score_percentage = score * 100

    return f"Final Should-I-breakup Score: {score_percentage:.2f}%" 

```

## Fill the Template

```python
# Dictionary of traits (score, weight)
positive_traits = {
    'Empathy': (1, 0.5),
    'Creativity': (3, 0.1),
    'Teamwork': (1, 0.1),
    'Leadership': (4, 0.3)
}

negative_traits = {
    'Procrastination': (2, 0.5),
    'Impulsiveness': (1, 0.5),
    'Mama\'s boy': (5, 0.0)
}

baseline = 1  # Value from 0 to 1 indicating happiness in the relationship

print(calculate_score(positive_traits, negative_traits, baseline))
```

| Score Range       | Interpretation           |
|------------------|--------------------------|
| $< 50\%$          | 🚩 Red flag. It's about time to retreat         |
| $50\% - 90\%$     | 🌝 Decent- could tip either way, so keep an eye on it  |
| $\geq 90\%$       | 🌲 Excellent performance! Keep growing the relationship     |
