# How Do the Traits Contribute to the Final “Should I Breakup” Score?

Run this notebook to see the importance of the positive and negative traits in the form of:
* A bubble chart where the size of the bubbles represent the value of the score multiplied by the respective weight
* A table that can be filtered in ascending and descending orders

## Reuse Your Data Points

```python
# Dictionary of traits (score, weight)
positive_traits = {
    'Empathy': (1, 0.5),
    'Creativity': (3, 0.1),
    'Teamwork': (1, 0.1),
    'Leadership': (4, 0.3)
}

negative_traits = {
    'Procrastination': (2, 0.5),
    'Impulsiveness': (1, 0.5),
    'Mama\'s boy': (5, 0.0)
}
```

## Import Libraries

```python
import plotly as ply
import numpy as np
if not hasattr(np, 'bool8'):
    np.bool8 = np.bool_
import plotly.express as px

import pandas as pd
from itables import show
```

```python
# A bubble chart that shows the importance of the traits

def plot_trait_bubbles(positive_traits, negative_traits):
    traits, scores, weights, types, sizes = [], [], [], [], []

    for trait, (score, weight) in positive_traits.items():
        traits.append(trait)
        scores.append(score)
        weights.append(weight)
        types.append('Positive')
        sizes.append(score * weight)

    for trait, (score, weight) in negative_traits.items():
        traits.append(trait)
        scores.append(score)
        weights.append(weight)
        types.append('Negative')
        sizes.append(score * weight)

    df = pd.DataFrame({
        'Trait': traits,
        'Score': scores,
        'Weight': weights,
        'Type': types,
        'Size': sizes
    })

    fig = px.scatter(
        df,
        x='Type',
        y='Score',
        size='Size',
        color='Type',
        text='Trait',
        hover_data=['Trait', 'Score', 'Weight', 'Size'],
        size_max=60,
        title='Trait Importance Bubble Chart (Weight × Score)',
    )

    fig.update_traces(textposition='top center')
    fig.show()

plot_trait_bubbles(positive_traits, negative_traits)
```

```python
# A table that can sort asc and desc

def display_itables(positive_traits, negative_traits):
    traits, scores, weights, types = [], [], [], []

    for trait, (score, weight) in positive_traits.items():
        traits.append(trait)
        scores.append(score)
        weights.append(weight)
        types.append('Positive')

    for trait, (score, weight) in negative_traits.items():
        traits.append(trait)
        scores.append(score)
        weights.append(weight)
        types.append('Negative')

    df = pd.DataFrame({
        'Type': types,
        'Trait': traits,
        'Score': scores,
        'Weight': weights

    })

    show(df, scrollY="200px", paging=False, columnDefs=[{"targets": "_all", "className": "dt-left"}])

display_itables(positive_traits, negative_traits)
```
